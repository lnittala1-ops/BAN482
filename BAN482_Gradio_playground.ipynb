## 2. Exploring Inputs
Gradio has many ways to accept input from a user. Let's look at the most common ones beyond just text.

### A. Sliders and Numbers
Great for adjusting parameters like age, price, or confidence thresholds.

```python
def calculate_tip(bill_amount, tip_percent):  # Define function to calculate total bill with tip
    total = bill_amount * (1 + tip_percent / 100)  # Calculate total by multiplying bill amount by (1 + tip percentage as decimal)
    return f"Total Bill: ${total:.2f}"  # Return formatted string with total bill rounded to 2 decimal places

demo = gr.Interface(  # Create a Gradio Interface to build the interactive UI
    fn=calculate_tip,  # Specify the function to execute when user submits input
    inputs=[  # Define list of input components for the interface
        gr.Number(label="Bill Amount ($)"),  # Number input for bill amount with dollar sign in label
        gr.Slider(minimum=0, maximum=30, step=1, value=15, label="Tip Percentage (%)")  # Slider input for tip: min=0, max=30, step=1, default=15
    ],  # Close the inputs list
    outputs="text",  # Specify output format as text
    title="💰 Tip Calculator"  # Set the title with calculator emoji
)  # Close the gr.Interface() constructor

demo.launch()  # Launch the interface to create a public shareable link
```

### B. Checkboxes, Radios, and Dropdowns
Used for categorical options.

```python
def order_pizza(size, toppings, extra_cheese):
    summary = f"Ordered a {size} pizza with {', '.join(toppings)}."
    if extra_cheese:
        summary += " (Plus extra cheese!)"
    return summary

demo = gr.Interface(
    fn=order_pizza,
    inputs=[
        gr.Radio(["Small", "Medium", "Large"], label="Size"),
        gr.CheckboxGroup(["Pepperoni", "Mushrooms", "Onions", "Olives"], label="Toppings"),
        gr.Checkbox(label="Extra Cheese?")
    ],
    outputs="text",
    title="🍕 Pizza Order UI"
)

demo.launch()
```

## 3. Handling Media (Images)
Gradio shines with media. You can input images and get images back.

```python
import numpy as np

def sepia_filter(input_img):
    # def sepia_filter(input_img):: This defines a Python function named sepia_filter that
    # takes one argument, input_img, which is expected to be a NumPy array representing an image.
    # Apply a simple sepia filter using numpy math

    sepia_filter = np.array([
        [0.393, 0.769, 0.189],
        [0.349, 0.686, 0.168],
        [0.272, 0.534, 0.131]
    ])
    # sepia_filter = np.array([...]): Inside the function, a 3x3 NumPy array (matrix) is defined.
    # This matrix represents the mathematical transformation (filter) to apply a sepia tone to an
    # image's RGB (Red, Green, Blue) color channels.

    sepia_img = input_img.dot(sepia_filter.T)
    # This line performs a matrix multiplication.
    # input_img (the original image's pixel data) is multiplied by the transpose (.T) of the
    # sepia_filter matrix. This effectively applies the sepia color transformation to each pixel.

    sepia_img /= sepia_img.max()
    # sepia_img /= sepia_img.max(): Image pixel values typically range from 0 to 255 (for 8-bit images)
    # or 0 to 1 (for floating-point representations). After applying the filter, some pixel values
    # might exceed these ranges. This line normalizes the pixel values in the sepia_img by dividing
    # them by the maximum value found in the processed image, ensuring they stay within a
    # valid range (0-1 in this case, assuming the input was also normalized or will be handled by Gradio).

    return sepia_img

    # return sepia_img: The function returns the NumPy array representing the image with the sepia filter applied.

demo = gr.Interface(
    fn=sepia_filter,
    inputs=gr.Image(label="Upload Image"), # Automatically handles upload
    outputs=gr.Image(label="Filtered Image"),
    title="🖼️ Instant Sepia Filter"
)

demo.launch()
```

## 4. Advanced Layouts with `gr.Blocks`
So far we used `gr.Interface`, which stacks everything vertically.
To create custom layouts (rows, columns, tabs), we use `gr.Blocks`. This is like building a web page.

```python
with gr.Blocks() as demo:
  # This is the entry point for creating custom, more complex Gradio user interfaces.
  # Unlike gr.Interface, gr.Blocks gives you fine-grained control over the layout
  # and interactivity of components.

    gr.Markdown("# 🏗️ Custom Layout Demo")
    # This displays a title for your UI using Markdown formatting.
    # The # indicates a top-level heading.

    with gr.Row():
    # This context manager is used to arrange components horizontally.
    # Any Gradio components defined within this block will appear side-by-side in the UI.

        # Everything in this 'with' block sits side-by-side
        inp_text = gr.Textbox(placeholder="Type something here...", label="Input")
        #  This creates a text input field where the user can type.
        # placeholder provides a hint, and label is the text displayed above the input box.

        out_text = gr.Textbox(label="Output")
        #  This creates a text display field. It's intended to show output and
        # is typically read-only from the user's perspective.

    btn = gr.Button("Go!", variant="primary")

    # Define interaction: When button is clicked, copy input to output reversed
    def reverse_text(text):
        return text[::-1]

    btn.click(fn=reverse_text, inputs=inp_text, outputs=out_text)
    # This is the core of the interactivity. It sets up an event listener:
    # When the btn (the "Go!" button) is clicked,
    # the reverse_text function (fn=reverse_text) is called,
    # the value from inp_text is passed as an argument to reverse_text,
    # and the return value of reverse_text is then used to update the out_text component.

demo.launch()
```

## 5. Tabs and Accordions
Great for organizing complex apps with many settings.

```python
with gr.Blocks() as demo:
    gr.Markdown("# 📂 Tabs Example")

    with gr.Tabs():
    # This creates a tabbed interface.
    # Components defined within this block will be organized into different tabs.

        with gr.TabItem("Basic Settings"):
          # This defines the first tab, labeled "Basic Settings".
          # Any components inside this 'with' block will appear on this tab.
            gr.Textbox(label="Name")
            # Creates a text input field for a name.
            gr.Slider(label="Age")
            # Creates a slider for selecting a numeric value, such as age.


        with gr.TabItem("Advanced Settings"):
          # This defines the second tab, labeled "Advanced Settings".
            gr.Checkbox(label="Developer Mode")
            # Provides a simple toggleable checkbox option.
            gr.ColorPicker(label="Theme Color")
            # Allows users to select a color from a color picker UI.

        with gr.TabItem("Help"):
          #  This defines the third tab, labeled "Help".
            gr.Markdown("This is a help tab with instructions.")

demo.launch()
```
